```python
from mip import *

NumTarefas = 6
NumEstações = 3
ConjuntoTarefas = range(1, NumTarefas+1)
ConjuntoEstações = range(1, NumEstações+1)

Tempo = { 1: 2, 2: 4, 3: 4, 4: 6, 5: 5, 6: 4}

modelo= Model(name = "Tarefas", sense = MINIMIZE, solver_name="HiGHS" )
x = {}
for j in ConjuntoTarefas:
  for k in ConjuntoEstações:
    x[(j,k)] = modelo.add_var(var_type=BINARY, name=f'É ou não produzida')


Ca = {}
for k in ConjuntoEstações:
  Ca[k] = modelo.add_var(var_type=CONTINUOUS, lb=0.0, name=f'tempo de ciclo')


modelo.objective = minimize(xsum(Ca[k] for k in ConjuntoEstações))

for k in ConjuntoEstações:
  modelo += xsum(x[(j,k)] for j in ConjuntoTarefas) == 1

for k in ConjuntoEstações:
  modelo += xsum(x[(j,k)] * Tempo[j] for j in ConjuntoTarefas) <= Ca[k]

modelo.optimize()
print('valor otimo:' , modelo.objective)
```

```python
from mip import *

NumProdutos = 5
ConjuntoProdutos = range(1,NumProdutos+1)

TempoProcessamento = {1: 64, 2: 53,3: 63,4: 99,5: 189}
P = TempoProcessamento

M = 100000

PrazoEntrega = {1: 100, 2: 70, 3: 150, 4: 601, 5: 118}
d = PrazoEntrega


print('verificação tempo de processamento:', P)
print('verificação  Prazo de entrega:', d)

modelo1 = Model(name="Atrasos", sense = MINIMIZE, solver_name="HiGHS" )

T = {}
for i in ConjuntoProdutos:
  T[i] = modelo1.add_var(var_type=CONTINUOUS, lb=0.0, name=f'Atraso')

y = {}
for i in ConjuntoProdutos:
  for j in ConjuntoProdutos:
    y[(i,j)] = modelo1.add_var(var_type=BINARY, name=f'é seguido?')

C = {}
for i in ConjuntoProdutos:
  C[i] = modelo1.add_var(var_type=CONTINUOUS, lb=0.0, name=f'somas dos processamentos')

Tmax = modelo1.add_var(var_type=CONTINUOUS, lb=0.0, name=f'Atraso')

modelo1.objective = minimize(Tmax )

for j in ConjuntoProdutos:
  modelo1 += xsum(y[(i,j)] for i in ConjuntoProdutos if i!=j ) == 1

for i in ConjuntoProdutos:
  modelo1 += xsum(y[(i,j)] for j in ConjuntoProdutos if j!=i ) == 1

for i in ConjuntoProdutos:
  for j in ConjuntoProdutos:

    if i!=j:

      modelo1 += C[j] >= C[i] + P[j] + M * ( y[(i,j)] - 1)

for i in ConjuntoProdutos:
  continue
  modelo1 += T[i] >= C[i] - d[i]

for i in ConjuntoProdutos:
  continue
  modelo1 += Tmax >= T[i]

print(modelo1)
modelo1.optimize()
print('valor otimo:', modelo1.objective_value)
for i in ConjuntoProdutos:
  print(C[i].x)
```
